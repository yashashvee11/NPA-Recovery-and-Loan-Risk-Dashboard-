```python
import pandas as pd
import numpy as np
```

<h3>Load the Data</h3>
First, load the generated CSV file into a pandas DataFrame:

```python
df_npa = pd.read_csv("NPA_Data.csv")
```

<h3>Inspect the Data</h3>
Check the structure and contents of your data:

```python
# Check for null or missing values
df_npa.isnull().sum()
```

```python
# Descriptive statistics for numerical columns
df_npa.describe()
```

<h3>Handle Missing Values</h3>
Identify and handle any missing or null values:

```python
# Display rows with missing values
df_npa[df_npa.isnull().any(axis=1)]
```

```python
# Fill missing restructuring dates with a placeholder value
df_npa['Restructuring_Date'] = df_npa['Restructuring_Date'].fillna('Not Restructured')
```

```python
# Drop rows with missing values (if necessary)
df_npa = df_npa.dropna()
```

<h3>Convert Data Types</h3>
Ensure that columns have the correct data types:

```python
# Convert Disbursement_Date and Restructuring_Date to datetime format
df_npa['Disbursement_Date'] = pd.to_datetime(df_npa['Disbursement_Date'])
df_npa['Restructuring_Date'] = pd.to_datetime(df_npa['Restructuring_Date'], errors='coerce')  # Handle invalid dates
```

```python
# Ensure numerical columns are correctly typed
numerical_columns = ['Loan_Amount', 'Overdue_Amount', 'Interest_Rate', 'Credit_Score', 'Annual_Income', 
                     'Provisioning_Amount', 'Recovery_Amount', 'Recovery_Percentage', 'Write_Off_Amount']
df_npa[numerical_columns] = df_npa[numerical_columns].apply(pd.to_numeric, errors='coerce')
```

<h3> Remove Duplicates</h3>
Check for and remove duplicate rows:

```python
# Check for duplicates
print(df_npa.duplicated().sum())
```

<h3>Ensure Data Consistency</h3>
Verify that categorical columns have valid values:

```python
# Check unique values in categorical columns
print(df_npa['Loan_Type'].unique())
print(df_npa['Industry'].unique())
print(df_npa['Region'].unique())
```

```python
# Replace incorrect or inconsistent values
df_npa['Region'] = df_npa['Region'].replace({'Central': 'North Central'})  # Example of renaming
```

<h3>Handle Outliers</h3>
Detect and handle outliers in numerical columns:

```python
# Identify outliers using the IQR method
Q1 = df_npa['Loan_Amount'].quantile(0.25)
Q3 = df_npa['Loan_Amount'].quantile(0.75)
IQR = Q3 - Q1

# Filter out outliers
df_npa = df_npa[(df_npa['Loan_Amount'] >= Q1 - 1.5 * IQR) & (df_npa['Loan_Amount'] <= Q3 + 1.5 * IQR)]
```

```python
# Save cleaned data
df_npa.to_csv('Cleaned_Npa_Data.csv', index=False)

print("Data cleaned and saved to 'cleaned_npa_data.csv'")
```
